### 📖 Overview
This lab demonstrates data acquisition from web portals. It introduces HTTP client communication via `requests`, HTML DOM parsing via `BeautifulSoup`, noise cleaning (script and style removal), and architectural scaffolding for enterprise scraping frameworks (Scrapy and Selenium).

### 🎯 Key Learning Objectives

- Extract target DOM nodes by tag names (`h1` through `h6`, `p`).
- Decompose noisy DOM nodes (`<script>`, `<style>`) to isolate text payloads.
- Test scraper logic across live targets (news articles and real-estate directories).

## Scrapping using bs4 and requests

```python
#Scraping Website using based on tags
import requests
from bs4 import BeautifulSoup

headers = {
    'User-Agent': 'Mozilla/5.0 (iPhone; CPU iPhone OS 8_3 like Mac OS X) AppleWebKit/600.1.4 (KHTML, like Gecko) Version/8.0 Mobile/12F70 Safari/600.1.4'
}
url = "https://www.yallakora.com/copa-america/2898/news/477889/%d9%83%d9%88%d8%a8%d8%a7-%d8%a3%d9%85%d8%b1%d9%8a%d9%83%d8%a7-2024-%d8%a7%d9%84%d8%a8%d8%b1%d8%a7%d8%b2%d9%8a%d9%84-%d8%aa%d9%86%d8%b9%d8%b4-%d8%a2%d9%85%d8%a7%d9%84-%d8%a7%d9%84%d8%aa%d8%a3%d9%87%d9%84-%d8%a8%d8%b1%d8%a8%d8%a7%d8%b9%d9%8a%d8%a9-%d8%a3%d9%85%d8%a7%d9%85-%d8%a8%d8%a7%d8%b1%d8%a7%d8%ac%d9%88%d8%a7%d9%8a-%d9%81%d9%8a%d8%af%d9%8a%d9%88-#HP_FeaturedArea"
# url="https://www.propertyfinder.eg/en/buy/properties-for-sale.html"
# url='https://en.wikipedia.org/wiki/IPhone_17_Pro'

# Send a GET request to the website
response = requests.get(url)
# print('respone',response.content)
# Check if the request was successful (status code 200)
if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html5lib') # lxml
    # print("Type",type(soup))
    # Extract all the headings (h1, h2, h3, h4, h5, h6)
    headings = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6'])
    # headings = soup.find('h1')

    # Print the text of each heading
    for heading in headings:
        print(heading.get_text())
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```
